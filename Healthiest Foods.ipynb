```python
# Included a new column called Food Type(P/A) for plant(P) and animal/animal source (A) foods
df = pd.read_csv("~/Library/Mobile Documents/com~apple~CloudDocs/Data Science Portfolio Projects/data where column was added(A include all animals and animal products incluing egg, yogurt, and fish too)/Top_100_Healthiest_Food_with_Plant_Animal_Column.csv")
df.head()
```

```python
"""Was there a difference in the types of foods considered the healthiest on the list?"""
```

```python
value_count = df['Food Type (P/A)'].value_counts()

print(value_count)
```

```python
value_count_normalize = df['Food Type (P/A)'].value_counts(normalize=True)

print(value_count_normalize)
```

```python
value_count_percentage = round(df['Food Type (P/A)'].value_counts(normalize=True)*100, 2)

print(value_count_percentage)
```

```python
Avg_P = value_count_percentage.get('P', 0)
Avg_A = value_count_percentage.get('A', 0)

print("Avg_P:", Avg_P)
print("Avg_A:", Avg_A)
```

```python
labels = ['Plant Source', 'Animal Source']
values = [Avg_P, Avg_A]

plt.pie(values, labels=labels, autopct='%1.2f%%', startangle=90)
plt.title('Percentage of Food Sources')
plt.show()
```

```python
plt.bar(labels, values, color=['green', 'orange'])
plt.ylabel('Percentage (%)')
plt.title('Plant vs Animal Food Sources')
plt.ylim(0, 100)
plt.show()
```

```python
# Chi-square test
observed = [91, 7]
expected = [49, 49]

stat, p_value = chisquare(observed, expected)

print("Chi-square:", stat)
print("P-value:", p_value)
```

```python
"""A chi-square goodness-of-fit test was conducted to determine whether plant-based and animal-based food sources occurred in equal proportions. The results showed a statistically significant difference between the two categories (χ² = 72.00, p < 0.001). Plant-based sources (92.86%) were substantially more prevalent than animal-based sources (7.14%). Therefore, the null hypothesis of equal proportions was rejected."""
```
